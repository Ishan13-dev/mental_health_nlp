```python
import pandas as pd 
import numpy as np

# Load the dataset  
df = pd.read_csv('../data/raw/mental_health.csv')

# look the first five rows 
display(df.head(5))
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns)
print("\nInfo:")
df.info()
```

```python
#Describe
df.describe(include="all")
```

```python
# Missing Values

df.isnull().sum()
```

```python
# Percentage of missing values 
df.isnull().mean() *100
```

```python
# Duplicates

df.duplicated().sum()
```

```python
target_col = "status"   
text_col = "statement"

print(df[target_col].value_counts())
```

```python
print(df[target_col].value_counts(normalize=True) * 100)
```

```python
import matplotlib.pyplot as plt 

df[target_col].value_counts().plot(kind='bar', color="skyblue" , edgecolor="violet")
plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
